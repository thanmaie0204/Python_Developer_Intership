```python
import sqlite3
from datetime import datetime

DATABASE = "tasks.db"

print("Task Management System")
print("Python environment is working!")
```

```python
def connect_db():
    return sqlite3.connect(DATABASE)

print("Database connection function created.")
```

```python
def create_table():
    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS tasks (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            title TEXT NOT NULL,
            description TEXT,
            status TEXT NOT NULL DEFAULT 'Pending',
            priority TEXT NOT NULL DEFAULT 'Medium',
            created_at TEXT NOT NULL
        )
    """)

    conn.commit()
    conn.close()

    print("Tasks table created successfully.")

create_table()
```

```python
def add_task(title, description, priority):
    if not title.strip():
        print("Task title cannot be empty.")
        return

    if priority not in ["Low", "Medium", "High"]:
        print("Invalid priority.")
        return

    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("""
        INSERT INTO tasks
        (title, description, status, priority, created_at)
        VALUES (?, ?, ?, ?, ?)
    """, (
        title,
        description,
        "Pending",
        priority,
        datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    ))

    conn.commit()
    conn.close()

    print("Task added successfully!")
```

```python
add_task(
    "Complete Python Internship",
    "Finish Task 3 implementation project",
    "High"
)
```

```python
def view_tasks():
    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute("SELECT * FROM tasks")
    tasks = cursor.fetchall()

    conn.close()

    if not tasks:
        print("No tasks found.")
        return

    for task in tasks:
        print("----------------------------")
        print("ID:", task[0])
        print("Title:", task[1])
        print("Description:", task[2])
        print("Status:", task[3])
        print("Priority:", task[4])
        print("Created:", task[5])

view_tasks()
```

```python
def update_task(task_id, new_status):
    valid_statuses = ["Pending", "In Progress", "Completed"]

    if new_status not in valid_statuses:
        print("Invalid status.")
        return

    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute(
        "UPDATE tasks SET status = ? WHERE id = ?",
        (new_status, task_id)
    )

    if cursor.rowcount == 0:
        print("Task not found.")
    else:
        print("Task status updated successfully!")

    conn.commit()
    conn.close()
```

```python
update_task(1, "Completed")
```

```python
def delete_task(task_id):
    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute(
        "DELETE FROM tasks WHERE id = ?",
        (task_id,)
    )

    if cursor.rowcount == 0:
        print("Task not found.")
    else:
        print("Task deleted successfully!")

    conn.commit()
    conn.close()
```

```python
add_task(
    "Practice Python",
    "Practice Python functions and loops",
    "Medium"
)
```

```python
def search_task(keyword):
    conn = connect_db()
    cursor = conn.cursor()

    cursor.execute(
        "SELECT * FROM tasks WHERE title LIKE ?",
        ('%' + keyword + '%',)
    )

    tasks = cursor.fetchall()
    conn.close()

    if not tasks:
        print("No matching tasks found.")
        return

    for task in tasks:
        print("----------------------------")
        print("ID:", task[0])
        print("Title:", task[1])
        print("Description:", task[2])
        print("Status:", task[3])
        print("Priority:", task[4])
        print("Created:", task[5])
```

```python
search_task("Internship")
```

```python
def main_menu():
    while True:
        print("\n==============================")
        print("     TASK MANAGEMENT SYSTEM")
        print("==============================")
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Update Task Status")
        print("4. Delete Task")
        print("5. Search Task")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            title = input("Enter task title: ")
            description = input("Enter description: ")
            priority = input("Enter priority (Low/Medium/High): ")

            add_task(title, description, priority)

        elif choice == "2":
            view_tasks()

        elif choice == "3":
            task_id = int(input("Enter task ID: "))
            status = input(
                "Enter status (Pending/In Progress/Completed): "
            )

            update_task(task_id, status)

        elif choice == "4":
            task_id = int(input("Enter task ID: "))
            delete_task(task_id)

        elif choice == "5":
            keyword = input("Enter keyword to search: ")
            search_task(keyword)

        elif choice == "6":
            print("Thank you for using Task Management System!")
            break

        else:
            print("Invalid choice. Please try again.")
```

```python
main_menu()
```
